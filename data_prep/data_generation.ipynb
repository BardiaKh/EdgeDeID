## Synthetic Medical Report Generation
Set `BACKEND`, `NUM_INPUTS`, and `OUTPUT_PATH` below, then run all cells.

```python
BACKEND = "haiku"       # "deepseek" | "haiku" | "hermes"
NUM_INPUTS = 500
OUTPUT_PATH = "../data/haiku/output_haiku.csv"
```

```python
from datagen import create_pipeline, generate_random_inputs

generator, gen_kwargs = create_pipeline(BACKEND, OUTPUT_PATH)
criteria = generate_random_inputs(NUM_INPUTS, **gen_kwargs)
generator.run(criteria)
```
